```python
# Import necessary Qiskit modules
from qiskit import QuantumCircuit
```

```python
# Create a quantum circuit with 2 qubits and no classical bits
qc = QuantumCircuit(2)

# Display the quantum circuit structure
print(qc)  # Qubits are labeled as "q_0", "q_1" (top to bottom)
```

```python
# Visualize the quantum circuit
qc.draw()  # Displays the circuit with qubits labeled as "q0", "q1" (top to bottom)
```

```python
# Create a quantum circuit with 2 qubits and 3 classical bits
qc = QuantumCircuit(2, 3)

# Visualize the quantum circuit
qc.draw('mpl')  # Displays qubits as single lines and classical bits as double lines
```

```python
# Define the number of qubits and classical bits using variables
num_qubits, num_bits = 4, 3

# Initialize the quantum circuit with specified qubits and bits
qc = QuantumCircuit(num_qubits, num_bits)

# Display the circuit as plain text
qc.draw('text')  # Provides a simple text-based visualization
```

```python
# Create a quantum circuit with 2 qubits and 3 classical bits
qc = QuantumCircuit(2, 3)

# Display the circuit as a diagram, showing the initial |0> states of qubits 
# and the 0 values for classical bits
qc.draw('mpl', initial_state=True)  
```

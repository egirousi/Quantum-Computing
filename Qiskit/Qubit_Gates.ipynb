```python
# Import necessary Qiskit module
from qiskit import QuantumCircuit

# Create a quantum circuit with 4 qubits (no classical bits needed)
qc = QuantumCircuit(4)

# Apply quantum gates to specific qubits

# X (Pauli-X) Gate - Also known as the NOT gate
# Swaps the |0⟩ and |1⟩ states, flipping the qubit
# Rotates the qubit 180° around the X-axis of the Bloch sphere
qc.x(0)  # Apply X gate to qubit 0

# Y (Pauli-Y) Gate
# Similar to X but also introduces a phase change
# Rotates the qubit 180° around the Y-axis of the Bloch sphere
qc.y(1)

# Z (Pauli-Z) Gate
# Leaves |0⟩ unchanged but flips the phase of |1⟩ (introduces a negative sign)
# Rotates the qubit 180° around the Z-axis of the Bloch sphere
qc.z(2)  # Apply Z gate to qubit 1

# Qubit 2 remains unchanged (no gates applied)

# X and Y gates applied sequentially on qubit 3
qc.x(3)  # Apply X gate to qubit 3 (flips state)

qc.y(3)  # Apply Y gate to qubit 3

# Visualize the circuit
qc.draw()
```

```python
# Create a quantum circuit with 2 qubits (no classical bits needed)
qc = QuantumCircuit(2)

# Apply quantum gates to specific qubits

# Qubit 0 remains unchanged (no gates applied)

# H (Hadamard) Gate - Creates superposition
# Rotates the qubit 180° around an axis between X and Z on the Bloch sphere
# Transforms |0⟩ into (|0⟩ + |1⟩)/√2 and |1⟩ into (|0⟩ - |1⟩)/√2
qc.h(1)  # Apply H gate to qubit 1

# Add a barrier for visualization
# A barrier does not affect computation, only improves circuit readability
qc.barrier()

# Visualize the circuit
qc.draw()
```

```python
# Create a quantum circuit with 3 qubits and 3 classical bits
qc = QuantumCircuit(3, 3)

# Apply quantum gates to specific qubits

# SWAP gate - Exchanges the quantum states of two qubits
# Swaps qubit No. 0 and qubit No. 1
qc.swap(0, 1)  # Swap states of qubit 0 and qubit 1

# Apply Hadamard (H) gate to qubit 0
# Transforms qubit 0 to an equal superposition of |0⟩ and |1⟩
qc.h(0)

# Controlled-SWAP (CSWAP) gate - Uses qubit No. 0 as control and qubits No. 1 and No. 2 as targets
# If qubit No. 0 is in state |1⟩, it swaps the states of qubits No. 1 and No. 2
qc.cswap(0, 1, 2)  # CSWAP with qubit 0 as control, qubits 1 and 2 as targets

# Barrier for visualization (doesn't affect computation)
qc.barrier()

# Swap qubits No. 1 and No. 0
# A swap gate is symmetric, so this is equivalent to swapping qubit 0 with qubit 1
qc.swap(1, 0)

# Apply another Controlled-SWAP gate 
qc.cswap(0, 1, 2)  # CSWAP with label for clarity

# Visualize the circuit
qc.draw()
```

```python
# Create a quantum circuit with 3 qubits and 3 classical bits
qc = QuantumCircuit(3, 3)

# Apply quantum gates to specific qubits

# Apply Hadamard (H) gate to qubit 0
# Transforms qubit 0 to an equal superposition of |0⟩ and |1⟩
qc.h(0)

# Add a barrier for visualization purposes
qc.barrier()

# Apply CNOT (Controlled-X) gate: Qubit 0 controls qubit 1
# If qubit 0 is in state |1⟩, apply an X gate (NOT gate) to qubit 1
qc.cx(0, 1)
qc.barrier()

# Apply another CNOT gate: Qubit 2 controls qubit 1
# If qubit 2 is in state |1⟩, apply an X gate (NOT gate) to qubit 1
qc.cx(2, 1)  # Same as .cx(2, 1)
qc.barrier()

# Apply Toffoli Gate (CCNOT): Qubits 0 and 1 control qubit 2
# If both qubits 0 and 1 are in state |1⟩, apply an X gate to qubit 2
qc.ccx(0, 1, 2)
qc.barrier()

# Apply Controlled-Z (CCZ) gate: Qubits 0 and 1 control qubit 2
# If both qubits 0 and 1 are in state |1⟩, apply a Z gate to qubit 2
# Note: The target (qubit 2) is not displayed in the circuit, so we use a label for clarity
qc.ccz(0, 1, 2)
qc.barrier()

# Apply Controlled-Y (CY) gate: Qubit 1 controls qubit 2
# If qubit 1 is in state |1⟩, apply a Y gate to qubit 2
qc.cy(1, 2)
qc.barrier()

# Apply Controlled-H (CH) gate: Qubit 1 controls qubit 0
# If qubit 1 is in state |1⟩, apply an H gate to qubit 0
qc.ch(1, 0)

# Visualize the circuit
qc.draw()
```
